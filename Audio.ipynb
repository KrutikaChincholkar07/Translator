```python
# Install required packages
!pip install -q openai-whisper torch soundfile pydub googletrans==4.0.0-rc1

# Imports
import whisper
from google.colab import files
from googletrans import Translator
from pydub import AudioSegment
import math
import os

# ---------------------------
# Step 1: Load Whisper Model
# ---------------------------
print("Loading Whisper large model...")
model = whisper.load_model("large")  # "large-v2" if available
print("Model loaded!")

# ---------------------------
# Step 2: Upload Audio
# ---------------------------
print("Upload your Urdu/Hindi audio (.wav/.mp3):")
uploaded = files.upload()
audio_file = list(uploaded.keys())[0]
print(f"Uploaded: {audio_file}")

# ---------------------------
# Step 3: Preprocess Audio (convert to mono, 16kHz WAV)
# ---------------------------
sound = AudioSegment.from_file(audio_file)
sound = sound.set_channels(1).set_frame_rate(16000)
processed_file = "processed_audio.wav"
sound.export(processed_file, format="wav")
print("Audio converted to mono 16kHz WAV")

# ---------------------------
# Step 4: Split long audio into 60-second chunks
# ---------------------------
segment_length_ms = 60 * 1000  # 60 seconds
audio_segments = [sound[i:i+segment_length_ms] for i in range(0, len(sound), segment_length_ms)]
print(f"Audio split into {len(audio_segments)} segment(s)")

# ---------------------------
# Step 5: Initialize Translator
# ---------------------------
translator = Translator()

# ---------------------------
# Step 6: Process each segment
# ---------------------------
raw_transcriptions = []
english_translations = []
hindi_translations = []

for idx, segment in enumerate(audio_segments):
    segment_file = f"segment_{idx}.wav"
    segment.export(segment_file, format="wav")

    # Transcribe and translate to English
    result = model.transcribe(segment_file, language="hi", task="translate")
    raw_text = result["text"].strip()

    # English text from Whisper (task="translate") might already be in English, but you can force translate if needed
    english_text = raw_text  # Whisper's translate task produces English
    hindi_text = translator.translate(english_text, src='en', dest='hi').text

    # Save outputs
    raw_transcriptions.append(raw_text)
    english_translations.append(english_text)
    hindi_translations.append(hindi_text)

    print(f"\nSegment {idx+1}/{len(audio_segments)} processed")

# ---------------------------
# Step 7: Combine all segments
# ---------------------------
full_raw = " ".join(raw_transcriptions)
full_english = " ".join(english_translations)
full_hindi = " ".join(hindi_translations)

print("\n\n Final Outputs\n")
print(" Raw Transcription (Urdu/Hindi Script):")
print(full_raw)
print("\n English Translation:")
print(full_english)
print("\n Hindi Translation (Devanagari Script):")
print(full_hindi)
```
